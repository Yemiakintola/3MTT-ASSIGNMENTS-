```python
# 1. Quiz Game

# Define the quiz questions and answers
quiz = [
    {"question": "What is the full meaning of OOP?", "answer": "Object Oriented Protocol"},
    {"question": "What do you call your father's brother?", "answer": "Uncle"},
    {"question": "Indexing in python pandas starts from what(give your answer in words)?", "answer": "zero"},
    {"question": "What is the first name of data science facilitator?", "answer": "Hannah"},
    {"question": "What do you call the process of fixing error in a code?", "answer": "debugging"}
]

# Initialize the score
score = 0

# Quiz loop
for index, item in enumerate(quiz, 1):
    user_response = input(f"Q{index}: {item['question']} ").strip().lower()
    correct_answer = item['answer'].strip().lower()

    if user_response == correct_answer:
        print("Correct!\n")
        score += 1
    else:
        print(f"Wrong! The correct answer is: {item['answer']}\n")

# Display final score
print(f"Quiz Over! You have scored {score} out of {len(quiz)}.")
```

```python
# Number Swapping Program

def get_number(prompt):
    # Get a valid number from the user.
    while True:
        try:
            return float(input(prompt))
        except ValueError:
            print("Invalid input! Please enter a valid number:")

# Get user input
first_num = get_number("Enter the first number: ")
second_num = get_number("Enter the second number: ")

# Display numbers before swapping
print(f"\nBefore swapping: first number = {first_num}, second number = {second_num}")

# Swap using tuple unpacking
first_num, second_num = second_num, first_num

# Display numbers after swapping
print(f"After swapping: first number = {first_num}, second number = {second_num}")
```

```python

# Number Checker Program

def is_prime(num):
    """Check if a number is prime."""
    if num <= 1:
        return False
    for i in range(2, int(num ** 0.5) + 1):
        if num % i == 0:
            return False
    return True

def is_armstrong(num):
    """Check if a number is an Armstrong number."""
    digits = str(num)
    power = len(digits)
    total = sum(int(digit) ** power for digit in digits)
    return total == num

def is_palindrome(num):
    """Check if a number is a Palindrome."""
    return str(num) == str(num)[::-1]

def main():
    """Main function to handle user input and display results."""
    try:
        number = int(input("Enter a positive integer: "))

        if number < 0:
            print("Invalid input! Please enter a positive integer.")
            return

        # Check for each condition
        prime = is_prime(number)
        armstrong = is_armstrong(number)
        palindrome = is_palindrome(number)

        # Display results
        print(f"\nResults for {number}:")

        if prime:
            print("- It is a Prime number.")
        if armstrong:
            print("- It is an Armstrong number.")
        if palindrome:
            print("- It is a Palindrome number.")

        if not (prime or armstrong or palindrome):
            print("- The number does not meet any special criteria.")

    except ValueError:
        print("Invalid input! Please enter a valid integer.")

# Run the program
main()
```

```python

# Fibonacci Sequence Generator

def generate_fibonacci(n):
    """Generate the first n terms of the Fibonacci sequence."""
    if n <= 0:
        print("Please enter a positive integer.")
        return

    fibonacci_sequence = [0, 1]  # Initializing the sequence with first two terms

    if n == 1:
        print("Fibonacci Sequence: [0]")
        return

    # Generate the sequence iteratively
    for _ in range(2, n):
        next_term = fibonacci_sequence[-1] + fibonacci_sequence[-2]
        fibonacci_sequence.append(next_term)

    # Display the result
    print(f"Fibonacci Sequence (first {n} terms): {fibonacci_sequence}")

def main():
    """Main function to handle user input."""
    try:
        n = int(input("Enter the number of terms to generate: "))
        generate_fibonacci(n)
    except ValueError:
        print("Invalid input! Please enter a valid integer.")

# Run the program
main()
```
